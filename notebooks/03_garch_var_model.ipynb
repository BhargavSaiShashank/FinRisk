# 03 GARCH & VaR Modeling
Fitting GARCH and GJR-GARCH models and estimating Value-at-Risk.

```
import pandas as pd
from src.garch_models import get_volatility_features
from src.var_estimation import generate_all_vars, calculate_risk_labels

log_returns = pd.read_csv('../data/processed/log_returns.csv', index_col=0, parse_dates=True)
returns = log_returns['^STOXX50E']

vol_df, res_garch, res_gjr = get_volatility_features(returns)
vars_df = generate_all_vars(returns, res_garch, res_gjr)

labels, c_threshold = calculate_risk_labels(returns, vars_df['VaR_GARCH_5%'])
print(f"Risk Threshold (c): {c_threshold:.5f}")

modeling_data = pd.concat([vol_df, vars_df, labels.rename('risk_label')], axis=1)
modeling_data.to_csv('../data/processed/modeling_results.csv')
print("Saved modeling results.")
```
