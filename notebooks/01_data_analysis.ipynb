# 01 Data Analysis & Visualization
Replicating the initial statistical analysis for Euro Stoxx 50 (2008-2025).

```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

log_returns = pd.read_csv('../data/processed/log_returns.csv', index_col=0, parse_dates=True)
primary = '^STOXX50E'

# Return Distribution
plt.figure(figsize=(10, 6))
sns.histplot(log_returns[primary], kde=True, bins=100)
plt.title('Return Distribution (^STOXX50E)')
plt.savefig('../results/plots/return_distribution.png')

# Volatility Clustering
plt.figure(figsize=(12, 6))
plt.plot(log_returns[primary], lw=0.5)
plt.title('Volatility Clustering Over Time')
plt.savefig('../results/plots/volatility_clustering.png')

# Correlation Matrix
plt.figure(figsize=(10, 8))
sns.heatmap(log_returns.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Asset Correlations')
plt.savefig('../results/plots/correlation_matrix.png')
```
